```python

# ============================================================
# CUSTOMER CHURN PREDICTION - COMPLETE EDA
# Google Colab | Telco Customer Churn Dataset
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
from IPython.display import display
import io

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

# 2. UPLOAD DATASET
print("Upload Telco Customer Churn CSV file:")
uploaded = files.upload()

file_name = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("\nDataset loaded successfully!")
print("Dataset Shape:", df.shape)

# 3. DATASET OVERVIEW
print("\n========== FIRST 5 ROWS ==========")
display(df.head())

print("\n========== DATASET INFO ==========")
df.info()

print("\n========== COLUMN NAMES ==========")
print(df.columns.tolist())

print("\n========== DUPLICATE ROWS ==========")
print(df.duplicated().sum())

# 4. DATA CLEANING AND MISSING VALUES
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"], errors="coerce"
)

print("\n========== MISSING VALUES ==========")
display(df.isnull().sum().to_frame("Missing Values"))

eda = df.dropna(subset=["TotalCharges"]).copy()

print("\nShape after handling missing TotalCharges:", eda.shape)

# 5. STATISTICAL SUMMARY
print("\n========== STATISTICAL SUMMARY ==========")
display(eda.describe().round(2))

print("\n========== CATEGORICAL SUMMARY ==========")
display(eda.describe(include="object"))

# 6. CHURN DISTRIBUTION
print("\n========== CHURN COUNT ==========")
display(eda["Churn"].value_counts().to_frame("Customers"))

print("\n========== CHURN PERCENTAGE ==========")
display(
    (eda["Churn"].value_counts(normalize=True) * 100)
    .round(2).to_frame("Percentage")
)

plt.figure(figsize=(6, 4))
sns.countplot(data=eda, x="Churn")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()

# 7. NUMERICAL FEATURE DISTRIBUTION
num_cols = ["tenure", "MonthlyCharges", "TotalCharges"]

eda[num_cols].hist(
    bins=30, figsize=(12, 5), edgecolor="black"
)
plt.suptitle("Numerical Feature Distributions")
plt.tight_layout()
plt.show()

# 8. NUMERICAL FEATURES VS CHURN
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for col, ax in zip(num_cols, axes):
    sns.boxplot(data=eda, x="Churn", y=col, ax=ax)
    ax.set_title(f"{col} vs Churn")

plt.tight_layout()
plt.show()

print("\n========== MEAN VALUES BY CHURN ==========")
display(eda.groupby("Churn")[num_cols].mean().round(2))

# 9. CATEGORICAL FEATURES VS CHURN
cat_cols = [
    "gender", "SeniorCitizen", "Partner", "Dependents",
    "PhoneService", "MultipleLines", "InternetService",
    "OnlineSecurity", "OnlineBackup", "DeviceProtection",
    "TechSupport", "StreamingTV", "StreamingMovies",
    "Contract", "PaperlessBilling", "PaymentMethod"
]

for col in cat_cols:
    churn_rate = (
        eda.groupby(col, dropna=False)["Churn"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .reset_index(name="Churn Rate (%)")
    )

    plt.figure(figsize=(8, 4))
    sns.barplot(data=churn_rate, x=col, y="Churn Rate (%)")
    plt.title(f"Churn Rate by {col}")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.show()

# 10. CONTRACT TYPE VS CHURN
print("\n========== CONTRACT VS CHURN (%) ==========")

contract_churn = pd.crosstab(
    eda["Contract"], eda["Churn"], normalize="index"
) * 100

display(contract_churn.round(2))

contract_churn.plot(
    kind="bar", stacked=True, figsize=(8, 5)
)
plt.title("Contract Type vs Churn")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# 11. TENURE GROUP ANALYSIS
eda["TenureGroup"] = pd.cut(
    eda["tenure"],
    bins=[-1, 12, 24, 48, 60, 72],
    labels=["0-12", "13-24", "25-48", "49-60", "61-72"]
)

tenure_churn = (
    eda.groupby("TenureGroup", observed=False)["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="Churn Rate (%)")
)

print("\n========== CHURN BY TENURE GROUP ==========")
display(tenure_churn.round(2))

sns.barplot(data=tenure_churn, x="TenureGroup", y="Churn Rate (%)")
plt.title("Churn Rate by Tenure Group")
plt.xlabel("Tenure (Months)")
plt.show()

# 12. MONTHLY CHARGES VS CHURN
plt.figure(figsize=(8, 5))
sns.histplot(
    data=eda, x="MonthlyCharges",
    hue="Churn", bins=30, kde=True
)
plt.title("Monthly Charges Distribution by Churn")
plt.show()

# 13. CORRELATION HEATMAP
plt.figure(figsize=(8, 5))
sns.heatmap(
    eda[num_cols].corr(),
    annot=True, cmap="coolwarm", fmt=".2f"
)
plt.title("Numerical Feature Correlation Heatmap")
plt.show()

# 14. FINAL EDA SUMMARY
total = len(eda)
churned = (eda["Churn"] == "Yes").sum()
retained = (eda["Churn"] == "No").sum()

print("\n========== FINAL EDA SUMMARY ==========")
print("Total Customers:", total)
print("Churned Customers:", churned)
print("Retained Customers:", retained)
print(f"Overall Churn Rate: {churned / total * 100:.2f}%")

print("\nAverage Tenure by Churn:")
display(eda.groupby("Churn")["tenure"].mean().round(2))

print("\nAverage Monthly Charges by Churn:")
display(eda.groupby("Churn")["MonthlyCharges"].mean().round(2))

print("\n========== EDA COMPLETED SUCCESSFULLY ==========")
```
